```python
import pystac_client
import planetary_computer
import rioxarray
import os
print("All imports OK!")
```

# Download GNATSGO rasters

```python
# soc raster folder
soc_folder = "D:\Databases\carbon\Soil_Rasters\gNATSGO"
```

```python
# Connect to the STAC catalog
catalog = pystac_client.Client.open(
    "https://planetarycomputer.microsoft.com/api/stac/v1",
    modifier=planetary_computer.sign_inplace
)
```

```python
# Illinois bounding box (minx, miny, maxx, maxy)
illinois_bbox = [-91.52, 36.96, -87.4, 42.6]
```

```python
# Search for the gNATSGO Raster collection
search = catalog.search(
    collections=["gnatsgo-rasters"],
    bbox=illinois_bbox
)
items = list(search.items())
n_items = len(items)
```

```python
for i in range(n_items):
    # Select the desired asset (e.g., soc0_100) and grab the signed URL
    asset_href = items[i].assets["soc0_100"].href

    # Read the raster directly or save it to your disk
    soc_data = rioxarray.open_rasterio(asset_href)

    # save to file
    soc_data.rio.to_raster(os.path.normpath(os.path.join(soc_folder,f"gNATSGO_soc0_100_raster{i}_g_m2.tif")))
```
