```python
import arcpy
from arcpy.sa import *
arcpy.CheckOutExtension("Spatial");
```

# Resample biomass raster

```python
soil_raster = r"D:/Databases/carbon/Soil_Rasters/soc_0_100cm_t_ha_mosaic_WGS1984_null_fill.tif"
```

```python
# make the soil raster the reference
ref_r     = arcpy.Raster(soil_raster)
ref_sr    = ref_r.spatialReference
ref_cell  = ref_r.meanCellWidth
```

```python
# set snap raster to reference — forces grid alignment
arcpy.env.snapRaster             = soil_raster
arcpy.env.outputCoordinateSystem = None
resample_method = "BILINEAR"  # continuous
```

```python
# resample
arcpy.management.Resample(
    in_raster=r"D:/Databases/carbon/WCMC_carbon_tonnes_per_ha/AGB_BGB_carbon_mosaic3_v1b_IL_BoxClip_NullZero.tif",
    out_raster=r"D:/Databases/carbon/WCMC_carbon_tonnes_per_ha/AGB_BGB_carbon_mosaic3_v1b_IL_BoxClip_NullZero_Resampled.tif",
    cell_size=str(ref_cell),
    resampling_type=resample_method
);
```

# Add soil and biomass rasters

```python
# create raster objects
ras1 = Raster(soil_raster)
ras2 = Raster(r"D:/Databases/carbon/WCMC_carbon_tonnes_per_ha/AGB_BGB_carbon_mosaic3_v1b_IL_BoxClip_NullZero_Resampled.tif")
```

```python
# perform additional
output_raster = ras1 + ras2
```

```python
# save output
output_raster.save(r"D:/Databases/carbon/Soil_Rasters/WCMC_Biomass_Plus_gNATSGO_SOC_100cm_t_ha.tif")
```
