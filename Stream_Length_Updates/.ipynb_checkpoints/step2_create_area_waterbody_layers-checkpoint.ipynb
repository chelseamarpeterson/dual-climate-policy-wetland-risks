```python
import arcpy
import os
arcpy.env.overwriteOutput = True
arcpy.env.outputCoordinateSystem = None
project_folder = r"C:/Users/Chels/OneDrive - University of Illinois - Urbana/Ch5_Sackett_Impacts/Wetland Mapping"
nhd_gdb = os.path.normpath(os.path.join(project_folder, "Illinois_NHD_Shapefiles.gdb"))
```

# Prepare area layer

## Project to albers

```python
# target coordinate system: NAD 1983 Contiguous USA Albers — WKID 102003
target_sr = arcpy.SpatialReference(102003)
```

```python
# starting layer in WGS1984
nhd_area_wgs1984 = os.path.normpath(os.path.join(nhd_gdb,"NHDArea_IL_Clip_WGS1984"))
```

```python
# projected layer in NAD 1983 Contiguous USA Albers
nhd_area_albers1983 = os.path.normpath(os.path.join(nhd_gdb,"NHDArea_IL_Clip_AlbersNAD1983"))
```

```python
arcpy.management.Project(
    in_dataset=nhd_area_wgs1984,
    out_dataset=nhd_area_albers1983,
    out_coor_system=target_sr,
    transform_method= "WGS_1984_(ITRF00)_To_NAD_1983" # "NAD_1983_To_WGS_1984_1"
);
```

## Select Streams/Rivers/Canals

```python
nhd_area_streams_rivers_canals = os.path.normpath(os.path.join(nhd_gdb,"NHDArea_StreamRiversCanals"))
```

```python
arcpy.conversion.ExportFeatures(
    in_features = nhd_area_albers1983,
    out_features = nhd_area_streams_rivers_canals,
    where_clause = "(ftype = 460) OR (ftype = 336)"
);
```

## Fill NHD layer type, hydro class, and buffer distance fields

### Layer type

```python
nhd_area_streams_rivers_canals = os.path.normpath(os.path.join(nhd_gdb,"NHDArea_StreamRiversCanals"))
```

```python
arcpy.management.AddField(nhd_area_streams_rivers_canals, "Type", "Text");
```

```python
arcpy.management.CalculateField(
    in_table=nhd_area_streams_rivers_canals,
    field="Type",
    expression="'Area'",
    expression_type="PYTHON3"
);
```

### Assume perennial hydro class

```python
nhd_area_streams_rivers_canals = os.path.normpath(os.path.join(nhd_gdb,"NHDArea_StreamRiversCanals"))
```

```python
arcpy.management.AddField(nhd_area_streams_rivers_canals, "Final_NHD_Hydro_Class", "LONG");
```

```python
arcpy.management.CalculateField(
    in_table=nhd_area_streams_rivers_canals,
    field="Final_NHD_Hydro_Class",
    expression="1",
    expression_type="PYTHON3"
);
```

### Buffer distance

```python
nhd_area_streams_rivers_canals = os.path.normpath(os.path.join(nhd_gdb,"NHDArea_StreamRiversCanals"))
```

```python
variable_buffer_rule = """
def VarBuff(Final_NHD_Hydro_Class):
    if Final_NHD_Hydro_Class == 1:
        return 20
    elif Final_NHD_Hydro_Class == 2:
        return 10
    elif Final_NHD_Hydro_Class == 3:
        return 5
"""
```

```python
arcpy.management.AddField(nhd_area_streams_rivers_canals, "Variable_Buffer", "LONG");
```

```python
arcpy.management.CalculateField(
    in_table=nhd_area_streams_rivers_canals,
    field="Variable_Buffer",
    expression = "VarBuff(!Final_NHD_Hydro_Class!)",
    expression_type="PYTHON3",
    code_block = variable_buffer_rule
);
```

# Prepare waterbody layers

## Project to albers

```python
nhd_waterbody_wgs1984 = os.path.normpath(os.path.join(nhd_gdb,"NHDWaterbody_IL_Clip_WGS1984"))
```

```python
nhd_waterbody_albers1983 = os.path.normpath(os.path.join(nhd_gdb,"NHDWaterbody_IL_AlbersNAD1983"))
```

```python
arcpy.management.Project(
    in_dataset=nhd_waterbody_wgs1984,
    out_dataset=nhd_waterbody_albers1983,
    out_coor_system=target_sr,
    transform_method= "WGS_1984_(ITRF00)_To_NAD_1983" # "NAD_1983_To_WGS_1984_1"
);
```

## Export perennial waterbodies

```python
nhd_waterbody_perennial = os.path.normpath(os.path.join(nhd_gdb,"NHDWaterbody_IL_Perennial"))
```

```python
perennial_codes = "(fcode = 39004) OR (fcode = 39009) OR (fcode = 39010) OR (fcode = 39011) OR (fcode = 39012) OR (fcode = 43615) OR (fcode = 43600 AND gnis_name IS NOT NULL) OR (fcode = 39000 AND areasqkm > 0.01)"
```

```python
arcpy.conversion.ExportFeatures(
    in_features = nhd_waterbody_albers1983,
    out_features = nhd_waterbody_perennial,
    where_clause = perennial_codes
);
```

```python
count1 = int(arcpy.management.GetCount(nhd_waterbody_albers1983)[0])
count2 = int(arcpy.management.GetCount(nhd_waterbody_perennial)[0])
print(f"Previous number of features: {count1}")
print(f"Selected number of features: {count2}")
```

## Export intermittent waterbodies

```python
nhd_waterbody_intermittent = os.path.normpath(os.path.join(nhd_gdb,"NHDWaterbody_IL_Intermittent"))
```

```python
intermittent_codes = "fcode IN (39004, 39009, 39010, 39011, 39012, 39012, 43615)) OR (fcode = 43600 AND gnis_name IS NULL) OR (fcode = 39000 AND areasqkm <= 0.01)"
```

```python
arcpy.conversion.ExportFeatures(
    in_features = nhd_waterbody_albers1983,
    out_features = nhd_waterbody_intermittent,
    where_clause = intermittent_codes
);
```

```python
count1 = int(arcpy.management.GetCount(nhd_waterbody_albers1983)[0])
count2 = int(arcpy.management.GetCount(nhd_waterbody_intermittent)[0])
print(f"Previous number of features: {count1}")
print(f"Selected number of features: {count2}")
```
