```python
import arcpy
import os
arcpy.env.overwriteOutput = True
arcpy.env.outputCoordinateSystem = None
project_folder = r"C:/Users/Chels/OneDrive - University of Illinois - Urbana/Ch5_Sackett_Impacts/Wetland Mapping"
nhd_gdb = os.path.normpath(os.path.join(project_folder, "Illinois_NHD_Shapefiles.gdb"))
```

# Project to Lambert

```python
# target coordinate system: NAD 1983 Contiguous USA Albers — WKID 102003
target_sr = arcpy.SpatialReference(102003)
```

```python
# starting NHD layer
step0_nhd = os.path.normpath(os.path.join(nhd_gdb, "NHDFlowline_Step0_IL_Clip_WGS1984"))
```

```python
# starting NHD layer
step1_nhd = os.path.normpath(os.path.join(nhd_gdb, "NHDFlowline_Step1_IL_Clip_AlbersNAD1983"))
```

```python
# project
arcpy.management.Project(
    in_dataset=step0_nhd,
    out_dataset=step1_nhd,
    out_coor_system=target_sr,
    transform_method= "WGS_1984_(ITRF00)_To_NAD_1983" # "NAD_1983_To_WGS_1984_1"
);
```

# Calculate flowline lengths

```python
# planar length
arcpy.management.CalculateGeometryAttributes(
    in_features=step1_nhd,
    geometry_property=[["Length_Km_Planar", "LENGTH"]],
    length_unit="KILOMETERS",
    coordinate_system=target_sr,  # NAD 1983 Albers
    coordinate_format="SAME_AS_INPUT"                  
);

# geodesic length
arcpy.management.CalculateGeometryAttributes(
    in_features=step1_nhd,
    geometry_property=[["Length_Km_Geodesic", "LENGTH_GEODESIC"]],
    length_unit="KILOMETERS",
    coordinate_system=target_sr,
    coordinate_format="SAME_AS_INPUT"
);
```

# Select non-isolated flowlines

```python
# non-isolated flowline layer
step2_nhd = os.path.normpath(os.path.join(nhd_gdb, "NHDFlowline_Step2_IL_NonIsolated"))
```

```python
arcpy.conversion.ExportFeatures(
    in_features = step1_nhd,
    out_features = step2_nhd,
    where_clause = "dnhydroseq NOT IN (0) OR uphydroseq NOT IN (0)"
);
```

```python
# compare feature counts
count1 = int(arcpy.management.GetCount(step1_nhd)[0])
count2 = int(arcpy.management.GetCount(step2_nhd)[0])
print(f"Isolate features: {count1 - count2}")
print(f"Selected number of features: {count2}")
```

# Assign hydrologic classification based on NHD attributes

```python
# make new field
arcpy.management.AddField(step2_nhd, "NHD_Hydro_Class", "LONG");
```

```python
# populate field with 0
arcpy.management.CalculateField(
    in_table=step2_nhd,
    field="NHD_Hydro_Class",
    expression="0",
    expression_type="PYTHON3"
);
```

## Assign by FCode

### Perennial (FCode = 46006)

```python
arcpy.management.MakeFeatureLayer(step2_nhd, "temp_layer");
arcpy.management.SelectLayerByAttribute(
    in_layer_or_view="temp_layer",
    where_clause="fcode = 46006",
    selection_type="NEW_SELECTION"
);
count = int(arcpy.management.GetCount("temp_layer")[0])
print(f"Selected: {count} features")

# calculate 1 only for selected features
arcpy.management.CalculateField(
    in_table="temp_layer",      # calculates on selection only
    field="NHD_Hydro_Class",
    expression="1",
    expression_type="PYTHON3"
);
arcpy.management.Delete("temp_layer");
```

### Intermittent (FCode = 46003)

```python
arcpy.management.MakeFeatureLayer(step2_nhd, "temp_layer");
arcpy.management.SelectLayerByAttribute(
    in_layer_or_view="temp_layer",
    where_clause="fcode = 46003",
    selection_type="NEW_SELECTION"
);
count = int(arcpy.management.GetCount("temp_layer")[0])
print(f"Selected: {count} features")

# calculate 1 only for selected features
arcpy.management.CalculateField(
    in_table="temp_layer",      # calculates on selection only
    field="NHD_Hydro_Class",
    expression="2",
    expression_type="PYTHON3"
);
arcpy.management.Delete("temp_layer");
```

### Ephemeral (FCode = 46007)

```python
arcpy.management.MakeFeatureLayer(step2_nhd, "temp_layer");
arcpy.management.SelectLayerByAttribute(
    in_layer_or_view="temp_layer",
    where_clause="fcode = 46007",
    selection_type="NEW_SELECTION"
);
count = int(arcpy.management.GetCount("temp_layer")[0])
print(f"Selected: {count} features")

# calculate 1 only for selected features
arcpy.management.CalculateField(
    in_table="temp_layer",      # calculates on selection only
    field="NHD_Hydro_Class",
    expression="3",
    expression_type="PYTHON3"
);
arcpy.management.Delete("temp_layer");
```

## Assign ditches with a name to perennial

```python
arcpy.management.MakeFeatureLayer(step2_nhd, "temp_layer");
arcpy.management.SelectLayerByAttribute(
    in_layer_or_view="temp_layer",
    where_clause="ftype = 336 AND gnis_name IS NOT NULL",
    selection_type="NEW_SELECTION"
);
count = int(arcpy.management.GetCount("temp_layer")[0])
print(f"Selected: {count} features")

# calculate 1 only for selected features
arcpy.management.CalculateField(
    in_table="temp_layer",      # calculates on selection only
    field="NHD_Hydro_Class",
    expression="1",
    expression_type="PYTHON3"
);
arcpy.management.Delete("temp_layer");
```

## Assign unknown streams (FCode = 46000) based on total drainage area 

### Perennial: Drainage area > 16.98 km2

```python
arcpy.management.MakeFeatureLayer(step2_nhd, "temp_layer");
arcpy.management.SelectLayerByAttribute(
    in_layer_or_view="temp_layer",
    where_clause="fcode = 46000 AND totdasqkm > 16.9772",
    selection_type="NEW_SELECTION"
);
count = int(arcpy.management.GetCount("temp_layer")[0])
print(f"Selected: {count} features")

# calculate 1 only for selected features
arcpy.management.CalculateField(
    in_table="temp_layer",      # calculates on selection only
    field="NHD_Hydro_Class",
    expression="1",
    expression_type="PYTHON3"
);
arcpy.management.Delete("temp_layer");
```

### Intermittent: Drainage area ∈ (0.60 km2, 16.98 km2]

```python
arcpy.management.MakeFeatureLayer(step2_nhd, "temp_layer");
arcpy.management.SelectLayerByAttribute(
    in_layer_or_view="temp_layer",
    where_clause="fcode = 46000 AND (totdasqkm > 0.5987 AND totdasqkm <= 16.9772)",
    selection_type="NEW_SELECTION"
);
count = int(arcpy.management.GetCount("temp_layer")[0])
print(f"Selected: {count} features")

# calculate 1 only for selected features
arcpy.management.CalculateField(
    in_table="temp_layer",      # calculates on selection only
    field="NHD_Hydro_Class",
    expression="2",
    expression_type="PYTHON3"
);
arcpy.management.Delete("temp_layer");
```

### Ephemeral: Drainage area <= 0.60 km2

```python
arcpy.management.MakeFeatureLayer(step2_nhd, "temp_layer");
arcpy.management.SelectLayerByAttribute(
    in_layer_or_view="temp_layer",
    where_clause="fcode = 46000 AND totdasqkm <= 0.5987",
    selection_type="NEW_SELECTION"
);
count = int(arcpy.management.GetCount("temp_layer")[0])
print(f"Selected: {count} features")

# calculate 1 only for selected features
arcpy.management.CalculateField(
    in_table="temp_layer",      # calculates on selection only
    field="NHD_Hydro_Class",
    expression="3",
    expression_type="PYTHON3"
);
arcpy.management.Delete("temp_layer");
```
