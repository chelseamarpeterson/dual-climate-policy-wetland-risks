```python
import arcpy
import os
arcpy.env.overwriteOutput = True
arcpy.env.outputCoordinateSystem = None
project_folder = r"C:/Users/Chels/OneDrive - University of Illinois - Urbana/Ch5_Sackett_Impacts/Wetland Mapping"
nhd_gdb = os.path.normpath(os.path.join(project_folder, "Illinois_NHD_Shapefiles.gdb"))
jd_water_gdb = os.path.normpath(os.path.join(project_folder, "Jurisdictional_Water_Shapefiles.gdb"))
```

# Erase waterbody and area polygons from flowlines (might not be necessary but repeating ArcPro steps)

```python
# buffer distances with 20 as a proxy for the variable buffer scenario
buffer_distances = [1, 10, 20, 100]
```

```python
# first erase area from flowlines
for b in buffer_distances:
    # original NHD
    arcpy.analysis.PairwiseErase(
        in_features = os.path.normpath(os.path.join(jd_water_gdb, f"NHDFlowline_Buffer{b}")),
        erase_features = os.path.normpath(os.path.join(jd_water_gdb, f"NHDArea_Buffer{b}")),
        out_feature_class = os.path.normpath(os.path.join(jd_water_gdb, f"NHDFlowlines_EraseArea{b}"))
    );
    
    # Brinkerhoff-updated
    arcpy.analysis.PairwiseErase(
        in_features = os.path.normpath(os.path.join(jd_water_gdb, f"NHDFlowline_Brinkerhoff_Buffer{b}")),
        erase_features = os.path.normpath(os.path.join(jd_water_gdb, f"NHDArea_Buffer{b}")),
        out_feature_class = os.path.normpath(os.path.join(jd_water_gdb, f"NHDFlowlines_Brinkerhoff_EraseArea{b}"))
    );
```

```python
# first erase waterbodies from flowlines without area
for b in buffer_distances:
    # original NHD
    arcpy.analysis.PairwiseErase(
        in_features = os.path.normpath(os.path.join(jd_water_gdb, f"NHDFlowlines_EraseArea{b}")),
        erase_features = os.path.normpath(os.path.join(jd_water_gdb, f"NHDWaterbody_Buffer{b}")),
        out_feature_class = os.path.normpath(os.path.join(jd_water_gdb, f"NHDFlowlines_EraseArea_Waterbodies{b}"))
    );
    
    # Brinkerhoff-updated
    arcpy.analysis.PairwiseErase(
        in_features = os.path.normpath(os.path.join(jd_water_gdb, f"NHDFlowlines_Brinkerhoff_EraseArea{b}")),
        erase_features = os.path.normpath(os.path.join(jd_water_gdb, f"NHDWaterbody_Brinkerhoff_Buffer{b}")),
        out_feature_class=os.path.normpath(os.path.join(jd_water_gdb, f"NHDFlowlines_Brinkerhoff_EraseArea_Waterbodies{b}"))
    );
```
