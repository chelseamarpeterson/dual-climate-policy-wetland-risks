### Total protected non-WOTUS wetland areas by protection category

```python
protection_types = ["county_ordinance","managed_biodiversity","managed_multuses"]
```

```python
protection_labels = ["County_Ordinances","Managed_Biodiversity","Managed_Multiple_Uses"]
```

```python
n_t = len(protection_types)
```

```python
type_area_csvs = [f"wetland_area_totals/step2_2024_nonwotus_{t}_area_county_wetland_area_totals.csv" for t in protection_types]
```

```python
# csv paths
type_area_paths = [os.path.normpath(os.path.join(county_summaries, f"{t}")) for t in type_area_csvs]
```

```python
# export paths
type_area_layers = [os.path.normpath(os.path.join(climate_impacts_gdb, f"Step2_2024_County_Unprotected_NonWOTUS_Wetland_Area_{t}")) for t in protection_labels]
```

```python
for t in range(n_t):
    # read in updated county area totals for 2024
    arcpy.management.MakeTableView(type_area_paths[t], 
                                   "temp_table")
    
    # make temp layer
    arcpy.management.MakeFeatureLayer(county_shapefile,
                                      "temp_layer")

    # join table
    arcpy.management.AddJoin(
        in_layer_or_view="temp_layer",
        in_field="NAME",  
        join_table="temp_table",
        join_field="NAME",   
        join_type="KEEP_ALL" 
    )
    
    # export
    arcpy.conversion.ExportFeatures(
        in_features="temp_layer",
        out_features=type_area_layers[t]
    )
    
    # delete temp layer
    arcpy.management.Delete("temp_layer")
```
