```python
import arcpy
import os
arcpy.env.overwriteOutput = True
arcpy.env.outputCoordinateSystem = None
project_folder = r"C:/Users/Chels/OneDrive - University of Illinois - Urbana/Ch5_Sackett_Impacts"
pad_gdb = os.path.normpath(os.path.join(project_folder, r"Databases/Protected Areas/PADUS4_0_State_IL_GDB/PADUS4_0_StateIL.gdb"))
census_gdb = os.path.normpath(os.path.join(project_folder, "Wetland Mapping/TIGER_Line_Census.gdb"))
```

# Project to target spatial reference

```python
# starting USGS protected areas layer
step0_pad = os.path.normpath(os.path.join(pad_gdb, "Step0_PADUS4_Combined_IL"))
step0_pad
```

```python
# target coordinate system: NAD 1983 Contiguous USA Albers — WKID 102003
target_sr = arcpy.SpatialReference(102003)
```

```python
# name of projected layer
step1_pad = os.path.normpath(os.path.join(pad_gdb, "Step1_PADUS4_Combined_IL_NAD1983Albers"))
step1_pad
```

```python
# project layer to target
arcpy.management.Project(
    in_dataset=step0_pad,
    out_dataset=step1_pad,
    out_coor_system=target_sr
);
```

```python
# verify output coordinate system
sr = arcpy.Describe(step1_pad).spatialReference
print(f"CRS:   {sr.name}")
print(f"WKID:  {sr.factoryCode}")
print(f"Units: {sr.linearUnitName}")
```

```python
# Clip to Illiois state boundary
```

```python
il_shape_nad1983_albers = os.path.normpath(os.path.join(census_gdb, "TIGER_2025_IL_NAD1983_Albers"))
il_shape_nad1983_albers
```

```python
# name of clipped layer
step2_pad = os.path.normpath(os.path.join(pad_gdb, "Step2_PADUS4_IL_Clip_AlbersNAD1983"))
step2_pad
```

```python
# clip US-PAD polygons to Illinois boundary
arcpy.analysis.PairwiseClip(
    in_features=step1_pad, 
    clip_features=il_shape_nad1983_albers, 
    out_feature_class=step2_pad
);
```

# Export layers with different GAP codes

```python
gap_codes = [1, 2, 3]
n_g = len(gap_codes)
```

```python
for g in gap_codes:
    out_gap_file = os.path.normpath(os.path.join(pad_gdb, f"Step4_PADUS4_CombIL_Pro{g}"))
    gap_clause = f"GAP_Sts = '{g}'"
    arcpy.conversion.ExportFeatures(
        in_features = step2_pad,
        out_features = out_gap_file,
        where_clause = gap_clause
    )
    count = int(arcpy.management.GetCount(out_gap_file)[0])
    print(f"Exported {count} features.")
```

# Use pairwise erase to prevent overlapping polygons and account for highest protection level

## Erase 1 from 2 and 3

```python
arcpy.analysis.PairwiseErase(
    in_features=os.path.normpath(os.path.join(pad_gdb, "Step4_PADUS4_CombIL_Pro2")), 
    erase_features=os.path.normpath(os.path.join(pad_gdb, "Step4_PADUS4_CombIL_Pro1")), 
    out_feature_class=os.path.normpath(os.path.join(pad_gdb, "Step5_PADUS4_CombIL_Pro2_Erase1"))
);
```

```python
arcpy.analysis.PairwiseErase(
    in_features=os.path.normpath(os.path.join(pad_gdb, "Step4_PADUS4_CombIL_Pro3")), 
    erase_features=os.path.normpath(os.path.join(pad_gdb, "Step4_PADUS4_CombIL_Pro1")), 
    out_feature_class=os.path.normpath(os.path.join(pad_gdb, "Step5_PADUS4_CombIL_Pro3_Erase1"))
);
```

## Erase 2 without 1 from 3 without 1

```python
arcpy.analysis.PairwiseErase(
    in_features=os.path.normpath(os.path.join(pad_gdb, "Step5_PADUS4_CombIL_Pro3_Erase1")), 
    erase_features=os.path.normpath(os.path.join(pad_gdb, "Step5_PADUS4_CombIL_Pro2_Erase1")), 
    out_feature_class=os.path.normpath(os.path.join(pad_gdb, "Step6_PADUS4_CombIL_Pro3_Erase12"))
);
```

## Merge and dissolve 1, 2 without 1, and 3 without 1 or 2

```python
merge_datasets = [
    os.path.normpath(os.path.join(pad_gdb, "Step4_PADUS4_CombIL_Pro1")),
    os.path.normpath(os.path.join(pad_gdb, "Step5_PADUS4_CombIL_Pro2_Erase1")),
    os.path.normpath(os.path.join(pad_gdb, "Step6_PADUS4_CombIL_Pro3_Erase12"))
]
merge_datasets
```

```python
arcpy.management.Merge(merge_datasets, os.path.normpath(os.path.join(pad_gdb, "Step8_PADUS4_CombIL_ProMerge")));
```

```python
arcpy.analysis.PairwiseDissolve(
    in_features=os.path.normpath(os.path.join(pad_gdb, "Step8_PADUS4_CombIL_ProMerge")),
    out_feature_class=os.path.normpath(os.path.join(pad_gdb, "Step8_PADUS4_CombIL_ProDissolve")),
    dissolve_field=["GAP_Sts","d_GAP_Sts"],
    multi_part="MULTI_PART"
);
```
