```python
# paths
repo_folder = r"C:/Users/Chels/OneDrive - University of Illinois - Urbana/Ch6_CASC_Project/Dual-Risk-Repo"
old_project_folder = r"C:/Users/Chels/OneDrive - University of Illinois - Urbana/Ch5_Sackett_Impacts"
census_gdb = os.path.normpath(os.path.join(old_project_folder, "Wetland Mapping/TIGER_Line_Census.gdb"))
climate_impacts_gdb = r"F:/Wetland_Climate_Impacts/Wetland Climate Impacts.gdb"
```

```python
# read in updated county area totals for 2024
totals_2024_df = pd.read_csv(os.path.normpath(os.path.join(repo_folder,"County_Summaries/wetland_area_totals/step2_2024_nonwotus_locally_unprotected_area_county_wetland_area_totals.csv")))
```

```python
# read in updated county area totals for 2024
differences_2024_df = pd.read_csv(os.path.normpath(os.path.join(repo_folder,"County_Summaries/wetland_area_differences/step2_nonwotus_locally_unprotected_area_county_wetland_area_differences.csv")))
```

```python
county_layer = os.path.normpath(os.path.join(census_gdb, "tl_2025_il_county_clipped_NAD1983_Albers"))
```

```python
# join totals to county dataframe
arcpy.management.MakeFeatureLayer(county_layer, "temp_layer");
arcpy.management.AddJoin(
    in_layer_or_view="temp_layer",
    in_field="NAME",  
    join_table=totals_2024_df,
    join_field="NAME",   
    join_type="KEEP_ALL" 
);
arcpy.conversion.ExportFeatures(
    in_features="temp_layer",
    out_features=os.path.normpath(os.path.join(climate_impacts_gdb, "Step2_Brinkerhoff_2024_County_UPNW_Totals"))
);
arcpy.management.Delete("temp_layer");
```
